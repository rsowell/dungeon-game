# Part 9: Put the Game in a Browser

Our next goal is to make the game playable by someone who does not have Python installed.

We will use **Pygbag**, which packages a Pygame project so it can run in a web browser.

Your project should look something like:

```text
web_text_adventure/
    main.py
    assets/
        move.wav
        pickup.wav
        danger.wav
        win.wav
```

Install Pygbag:

```bash
python -m pip install --upgrade pygbag
```

From the directory containing the project folder, run:

```bash
python -m pygbag web_text_adventure
```

Pygbag will start a local web server.

Open the address it gives you in a browser.

# Part 10: What Makes This Work on the Web?

Near the end of `main.py`, find:

```python
async def main():
```

Then find this line inside the game loop:

```python
await asyncio.sleep(0)
```

You do not need to understand asynchronous programming yet.

For now, think of this line as the program briefly saying:

> “Browser, you can have control back now.”

A desktop Python program can sometimes sit inside its own loop and control everything.

A program running inside a browser needs to cooperate with the browser.

That small change is what allows our Pygame loop to work in the web environment.
